Analyse un joueur pour comprendre un peu Plus Data


````python
# Analyse par joueur spécifique X hnaya
def analyze_player(events, player_name):
    player_events = events[events['player'] == player_name]

    print(f"\nAnalyse de {player_name}")
    print(f"{'='*60}")
    print(f"Total événements: {len(player_events)}")
    print("\nTypes d'événements:")
    print(player_events['type'].value_counts())

    # aprés ndiro Statistiques de passes pour un joueur taux de réussite
    passes = player_events[player_events['type'] == 'Pass']
    if len(passes) > 0:
        print(f"\nStatistiques de passes:")
        print(f"  Total: {len(passes)}")
        print(f"  Réussies: {passes['pass_outcome'].isna().sum()}")
        print(f"  Taux de réussite: {passes['pass_outcome'].isna().sum()/len(passes)*100:.1f}%")

    return player_events

# Exemple d'utilisation
if len(events) > 0:
    player_name = events['player'].dropna().iloc[0]
    player_data = analyze_player(events, player_name)
````

````python
# Filtrer les joueurs par position
def get_players_by_position(lineups, position_filter):
    all_players = []

    for team_name, players in lineups.items():
        for idx, player in players.iterrows():
            positions = player.get('positions', [])
            if isinstance(positions, list) and len(positions) > 0:
                main_position = positions[0].get('position', '')
                if position_filter.upper() in main_position:
                    all_players.append({
                        'team': team_name,
                        'name': player['player_name'],
                        'nickname': player.get('player_nickname', ''),
                        'position': main_position,
                        'jersey': player.get('jersey_number', '')
                    })

    return pd.DataFrame(all_players)

# Exemples:
# Gardiens
goalkeepers = get_players_by_position(lineups, 'GK')
print("Gardiens:")
print(goalkeepers)

# Attaquants
forwards = get_players_by_position(lineups, 'Forward')
print("\nAttaquants:")
print(forwards)
````

````python
# Statistiques agrégrégation par position par position pour comprendre la golabilité des données
def stats_by_position(events, lineups):
    # Créer un mapping joueur -> position
    player_positions = {}
    for team_name, players in lineups.items():
        for idx, player in players.iterrows():
            positions = player.get('positions', [])
            if isinstance(positions, list) and len(positions) > 0:
                main_pos = positions[0].get('position', 'Unknown')
                player_positions[player['player_name']] = main_pos

    # Ajouter la position aux événements
    events['position'] = events['player'].map(player_positions)

    # Statistiques par position
    position_stats = events.groupby('position').agg({
        'id': 'count',  # Total événements
        'type': lambda x: x.value_counts().to_dict()
    }).reset_index()

    return position_stats, events

pos_stats, events_with_pos = stats_by_position(events, lineups)
print(pos_stats)
````

````python
# ÉTAPE 2: RÉCUPÉRER LES COMPOSITIONS (LINEUPS)
# ============================================

lineups = sb.lineups(match_id=match_id)
print("\n" + "="*80)
print("📋 COMPOSITIONS D'ÉQUIPE")
print("="*80)
````

````python
all_positions = set()
all_players_data = []

for team_name, players in lineups.items():
    print(f"\n{'='*80}")
    print(f"Équipe: {team_name}")
    print(f"{'='*80}")

    for idx, player in players.iterrows():
        positions = player.get('positions', [])
        if isinstance(positions, list):
            for pos_info in positions:
                if isinstance(pos_info, dict):
                    position = pos_info.get('position', 'Unknown')
                    all_positions.add(position)
                    player_data = {
                        'team': team_name,
                        'player_id': player.get('player_id', ''),
                        'player_name': player.get('player_name', ''),
                        'player_nickname': player.get('player_nickname', ''),
                        'jersey_number': player.get('jersey_number', ''),
                        'position': position,
                        'position_from': pos_info.get('from', ''),
                        'position_to': pos_info.get('to', ''),
                        'position_start_reason': pos_info.get('start_reason', '')
                    }
                    all_players_data.append(player_data)
# Afficher les joueurs de cette équipe
    team_df = pd.DataFrame([p for p in all_players_data if p['team'] == team_name])
    print(team_df[['player_name', 'jersey_number', 'position']].to_string(index=False))
````

Mapper les données ici pour chaque postes

````python
# Créer un dictionnaire player -> position
player_positions = {}
for player_data in all_players_data:
    player_positions[player_data['player_name']] = player_data['position']

# Ajouter la position aux événements
events['position'] = events['player'].map(player_positions)

print("\n" + "="*80)
print("🎯 MÉTRIQUES DISPONIBLES PAR POSTE")
print("="*80)
````

````python
# Pour chaque poste, montrer les métriques disponibles
for position in sorted(all_positions):
    position_events = events[events['position'] == position]

    if len(position_events) > 0:
        print(f"\n{'='*80}")
        print(f"📍 POSTE: {position}")
        print(f"{'='*80}")
        print(f"Nombre d'événements: {len(position_events)}")
        print(f"Joueurs à ce poste: {position_events['player'].nunique()}")

        # Types d'événements pour ce poste
        print("\n📊 Types d'événements:")
        pos_event_types = position_events['type'].value_counts()
        print(pos_event_types.to_string())

        # Colonnes non-nulles pour ce poste (métriques disponibles)
        print("\n📋 Métriques avec données (colonnes non-vides):")
        non_null_cols = []
        for col in events.columns:
            if position_events[col].notna().any():
                non_null_count = position_events[col].notna().sum()
                non_null_cols.append(f"  • {col} ({non_null_count} valeurs)")

        for metric in sorted(non_null_cols)[:30]:  # Afficher les 30 premières
            print(metric)

        if len(non_null_cols) > 30:
            print(f"  ... et {len(non_null_cols) - 30} autres métriques")
````

````python
# ÉTAPE 6: CRÉER UN RÉSUMÉ GLOBAL

print("\n" + "="*80)
print("📊 RÉSUMÉ GLOBAL: MÉTRIQUES PAR CATÉGORIE DE POSTE")
print("="*80)

# Catégoriser les postes
position_categories = {
    'Gardiens': [],
    'Défenseurs': [],
    'Milieux': [],
    'Attaquants': []
}

for pos in all_positions:
    if 'Goalkeeper' in pos or 'Keeper' in pos:
        position_categories['Gardiens'].append(pos)
    elif 'Back' in pos or 'Center Back' in pos or 'Wing Back' in pos:
        position_categories['Défenseurs'].append(pos)
    elif 'Midfield' in pos or 'Midfielder' in pos:
        position_categories['Milieux'].append(pos)
    elif 'Forward' in pos or 'Wing' in pos or 'Striker' in pos:
        position_categories['Attaquants'].append(pos)

for category, positions in position_categories.items():
    if positions:
        print(f"\n🎯 {category.upper()}:")
        for pos in sorted(positions):
            count = len(events[events['position'] == pos])
            print(f"   • {pos}: {count} événements")
````
